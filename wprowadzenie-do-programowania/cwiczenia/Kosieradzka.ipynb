## 2. Typy danych, zmienne, operatory, wejście/wyjście

### Zadanie 2.1
**Polecenie:** Pobierz od użytkownika dwie liczby i wypisz:
- ich sumę
- różnicę
- iloczyn
- wynik dzielenia

````python
# Zadanie 2.1 - Kalkulator podstawowy
liczba1 = float(input("Podaj pierwszą liczbę: "))
liczba2 = float(input("Podaj drugą liczbę: "))

suma = liczba1 + liczba2
roznica = liczba1 - liczba2
iloczyn = liczba1 * liczba2

# Sprawdzamy dzielenie przez zero
if liczba2 != 0:
    dzielenie = liczba1 / liczba2
    print(f"\nWyniki operacji:")
    print(f"Suma: {liczba1} + {liczba2} = {suma}")
    print(f"Różnica: {liczba1} - {liczba2} = {roznica}")
    print(f"Iloczyn: {liczba1} * {liczba2} = {iloczyn}")
    print(f"Dzielenie: {liczba1} / {liczba2} = {dzielenie}")
else:
    print(f"\nWyniki operacji:")
    print(f"Suma: {liczba1} + {liczba2} = {suma}")
    print(f"Różnica: {liczba1} - {liczba2} = {roznica}")
    print(f"Iloczyn: {liczba1} * {liczba2} = {iloczyn}")
    print(f"Dzielenie: Nie można dzielić przez zero!")
````

## 3. Instrukcje warunkowe

### Zadanie 3.1
**Polecenie:** Napisz program, który sprawdza, czy podana liczba jest:
- dodatnia
- ujemna
- równa zero

````python
# Zadanie 3.1 - Sprawdzenie znaku liczby
liczba = float(input("Podaj liczbę: "))

if liczba > 0:
    print(f"Liczba {liczba} jest dodatnia.")
elif liczba < 0:
    print(f"Liczba {liczba} jest ujemna.")
else:
    print(f"Liczba jest równa zero.")
````

### Zadanie 3.2
**Polecenie:** Napisz program sprawdzający, czy liczba jest parzysta.

````python
# Zadanie 3.2 - Sprawdzenie parzystości
liczba = int(input("Podaj liczbę całkowitą: "))

if liczba % 2 == 0:
    print(f"Liczba {liczba} jest parzysta.")
else:
    print(f"Liczba {liczba} jest nieparzysta.")
````

### Zadanie 3.3
**Polecenie:** Napisz program, który pobiera wynik testu (0-100) i wypisuje ocenę według skali:
- 90-100: 5
- 75-89: 4
- 60-74: 3
- 0-59: 2

````python
# Zadanie 3.3 - System oceniania
wynik = float(input("Podaj wynik testu (0-100): "))

# Walidacja zakresu
if wynik < 0 or wynik > 100:
    print("Błąd! Wynik musi być w zakresie 0-100.")
else:
    if wynik >= 90:
        ocena = 5
    elif wynik >= 75:
        ocena = 4
    elif wynik >= 60:
        ocena = 3
    else:
        ocena = 2
    
    print(f"Wynik testu: {wynik} punktów -> Ocena: {ocena}")
````

## 4. Pętle (for, while)

### Zadanie 4.1
**Polecenie:** Wyświetl liczby od 1 do 20.

````python
# Zadanie 4.1 - Wyświetlanie liczb od 1 do 20
print("Liczby od 1 do 20:")
for i in range(1, 21):
    print(i, end=" ")
print()  # Nowa linia na końcu
````

### Zadanie 4.2
**Polecenie:** Policz sumę liczb od 1 do n (n podaje użytkownik).

````python
# Zadanie 4.2 - Suma liczb od 1 do n
n = int(input("Podaj liczbę n: "))

suma = 0
for i in range(1, n + 1):
    suma += i

print(f"Suma liczb od 1 do {n} wynosi: {suma}")

# Sprawdzenie za pomocą wzoru matematycznego: suma = n * (n + 1) / 2
suma_matematyczna = n * (n + 1) // 2
print(f"Sprawdzenie (wzór matematyczny): {suma_matematyczna}")
````

### Zadanie 4.3
**Polecenie:** Narysuj prostokąt z gwiazdek o wymiarach podanych przez użytkownika, np.:  
```
Kod
****
****
****
```

````python
# Zadanie 4.3 - Prostokąt z gwiazdek
szerokosc = int(input("Podaj szerokość prostokąta: "))
wysokosc = int(input("Podaj wysokość prostokąta: "))

print("\nKod")
for i in range(wysokosc):
    print("*" * szerokosc)
````

### Zadanie 4.4
**Polecenie:** Narysuj prostą "choinkę" ("Zapytaj liczbę" od 1 do 20).

````python
# Zadanie 4.4 - Choinka z gwiazdek
wysokosc = int(input("Podaj wysokość choinki (1-20): "))

# Walidacja zakresu
if wysokosc < 1 or wysokosc > 20:
    print("Błąd! Wysokość musi być w zakresie 1-20.")
else:
    print("\nChoinka:")
    for i in range(1, wysokosc + 1):
        # Wyświetl spacje do wyśrodkowania
        spacje = " " * (wysokosc - i)
        # Wyświetl gwiazdki
        gwiazdki = "*" * (2 * i - 1)
        print(spacje + gwiazdki)
    
    # Opcjonalnie: pień choinki
    pien_spacje = " " * (wysokosc - 1)
    print(pien_spacje + "*")
````
